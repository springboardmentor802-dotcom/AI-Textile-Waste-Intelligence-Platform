```python
from PIL import Image
import matplotlib.pyplot as plt
import os

defects_path = '../Dataset/fabric_defect_dataset'

categories = ['defect free', 'stain', 'hole', 'lines', 'horizontal', 'Vertical']
fig, axes = plt.subplots(1, len(categories), figsize=(18, 3))

for ax, category in zip(axes, categories):
    folder_path = os.path.join(defects_path, category)
    image_extensions = ('.jpg', '.jpeg', '.png', '.bmp')
    sample_files = [f for f in os.listdir(folder_path) if f.lower().endswith(image_extensions)]
    img = Image.open(os.path.join(folder_path, sample_files[0]))
    ax.imshow(img)
    ax.set_title(category, fontsize=9)
    ax.axis('off')

plt.show()
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import os

defects_path = '../Dataset/fabric_defect_dataset'

categories = ['defect free', 'stain', 'hole', 'lines', 'horizontal', 'Vertical']
fig, axes = plt.subplots(1, len(categories), figsize=(18, 3))

for ax, category in zip(axes, categories):
    folder_path = os.path.join(defects_path, category)
    image_extensions = ('.jpg', '.jpeg', '.png', '.bmp')
    sample_files = [f for f in os.listdir(folder_path) if f.lower().endswith(image_extensions)]
    img = Image.open(os.path.join(folder_path, sample_files[0]))
    ax.imshow(img)
    ax.set_title(category, fontsize=9)
    ax.axis('off')

plt.show()
```

```python
categories = ['defect free', 'stain', 'hole', 'lines', 'horizontal', 'Vertical']
grayscale_categories = ['defect free']  # known single-channel folder

fig, axes = plt.subplots(1, len(categories), figsize=(18, 3))

for ax, category in zip(axes, categories):
    folder_path = os.path.join(defects_path, category)
    files = [f for f in os.listdir(folder_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    img = Image.open(os.path.join(folder_path, files[0]))

    if category in grayscale_categories:
        ax.imshow(img, cmap='gray')
    else:
        ax.imshow(img)

    ax.set_title(category, fontsize=9)
    ax.axis('off')

plt.show()
```
