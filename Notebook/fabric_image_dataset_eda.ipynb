```python
import os

dataset_path = '../Dataset/fabric_image_dataset'

for item in os.listdir(dataset_path):
    print(item)
```

```python
# count images per class
for folder in sorted(os.listdir(dataset_path)):
    folder_path = os.path.join(dataset_path, folder)
    if os.path.isdir(folder_path):
        num_images = len(os.listdir(folder_path))
        print(f"{folder}: {num_images} images")
```

```python
# total count across everything
total = 0
for folder in os.listdir(dataset_path):
    folder_path = os.path.join(dataset_path, folder)
    if os.path.isdir(folder_path):
        total += len(os.listdir(folder_path))
print(f"Total images: {total}")
```

```python
from PIL import Image
import matplotlib.pyplot as plt

sample_folder = '../Dataset/fabric_image_dataset/Cotton'
sample_files = os.listdir(sample_folder)[:5]

fig, axes = plt.subplots(1, 5, figsize=(15, 3))
for ax, filename in zip(axes, sample_files):
    img = Image.open(os.path.join(sample_folder, filename))
    ax.imshow(img)
    ax.set_title(filename, fontsize=8)
    ax.axis('off')
plt.show()
```

```python
from PIL import Image
import matplotlib.pyplot as plt

sample_folder = '../Dataset/fabric_image_dataset/Cotton'

image_extensions = ('.jpg', '.jpeg', '.png', '.bmp')
sample_files = [
    f for f in os.listdir(sample_folder)
    if f.lower().endswith(image_extensions)
][:5]

fig, axes = plt.subplots(1, 5, figsize=(15, 3))
for ax, filename in zip(axes, sample_files):
    img = Image.open(os.path.join(sample_folder, filename))
    ax.imshow(img)
    ax.set_title(filename, fontsize=8)
    ax.axis('off')
plt.show()
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import os

classes_to_show = ['Cotton', 'Denim', 'Silk', 'Wool', 'Polyester']
base_path = '../Dataset/fabric_image_dataset'

fig, axes = plt.subplots(1, len(classes_to_show), figsize=(15, 3))

for ax, fabric_class in zip(axes, classes_to_show):
    class_folder = os.path.join(base_path, fabric_class)
    first_swatch = sorted(os.listdir(class_folder))[0]
    image_path = os.path.join(class_folder, first_swatch, 'im_1.png')
    img = Image.open(image_path)
    ax.imshow(img)
    ax.set_title(fabric_class, fontsize=10)
    ax.axis('off')

plt.show()
```
