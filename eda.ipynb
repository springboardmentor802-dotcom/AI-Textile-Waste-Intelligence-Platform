```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Tumhare folder structure ke hisaab se sahi path:
base_dir = r'AI-Textile-Waste-Intelligence-Platform/Fabric_data-LAPTOP-3QGF9V05/train'

# Categories (cotton, silk, wool) ko list karo
categories = os.listdir(base_dir)

# Har category mein kitni images hain, count karo
data_counts = {cat: len(os.listdir(os.path.join(base_dir, cat))) for cat in categories}

# Console par print karo
for cat, count in data_counts.items():
    print(f"{cat}: {count} images")

# Graph banao
plt.figure(figsize=(10, 5))
sns.barplot(x=list(data_counts.keys()), y=list(data_counts.values()))
plt.title('Distribution of Textile Waste Classes')
plt.xlabel('Category')
plt.ylabel('Number of Images')
plt.show()
```
