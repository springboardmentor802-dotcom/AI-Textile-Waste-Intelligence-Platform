```python


def calculate_recyclability_score(material_info):

    score = 0

    # Reusable
    if material_info["reusable"]:
        score += 30

    # Biodegradable
    if material_info["biodegradable"] == True:
        score += 30
    elif material_info["biodegradable"] == "Partially":
        score += 15

    # Recycling Method
    if material_info["recycling_method"] == "Mechanical Recycling":
        score += 20
    elif material_info["recycling_method"] == "Chemical Recycling":
        score += 15
    elif material_info["recycling_method"] == "Reuse / Upcycling":
        score += 20
    elif material_info["recycling_method"] == "Specialized Recycling":
        score += 10

    # Environmental Impact
    if material_info["environmental_impact"] == "Low":
        score += 20
    elif material_info["environmental_impact"] == "Medium":
        score += 10

    return score
```

```python


def categorize_waste(score):

    if score >= 80:
        return "Highly Recyclable"

    elif score >= 50:
        return "Moderately Recyclable"

    else:
        return "Low Recyclability"
```

```python


def generate_recommendation(predicted_class):

    predicted_class = str(predicted_class).zfill(3)

    material = material_database[predicted_class]

    score = calculate_recyclability_score(material)

    category = categorize_waste(score)

    report = {

        "Predicted Class": predicted_class,

        "Material": material["material"],

        "Material Type": material["material_type"],

        "Recyclability Score": score,

        "Waste Category": category,

        "Recommended Recycling Method":
            material["recycling_method"],

        "Environmental Impact":
            material["environmental_impact"]

    }

    return report
```

```python


report = generate_recommendation("001")

for key, value in report.items():
    print(f"{key}: {value}")
```

```python


sample_classes = [
    "001",
    "002",
    "005",
    "007",
    "009",
    "010"
]

for fabric_class in sample_classes:

    report = generate_recommendation(fabric_class)

    print("=" * 60)

    for key, value in report.items():
        print(f"{key}: {value}")

    print()
```

```python

reports = []

for fabric_class in material_database.keys():

    reports.append(
        generate_recommendation(fabric_class)
    )

report_df = pd.DataFrame(reports)

report_df
```

```python


report_path = MODELS_PATH / "recyclability_report.csv"

report_df.to_csv(
    report_path,
    index=False
)

print("Report saved successfully")
print(report_path)
```
